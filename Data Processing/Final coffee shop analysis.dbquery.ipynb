```sql

--------------------------------------------------
---Bright Coffee Shop Coding
--------------------------------------------------

---1.Running the entire table
select * 
from `workspace`.`default`.`case1`;

-------------------------------------------------
---2.Checking the date range
SELECT
     MIN(transaction_date) AS Minimum_date,
     MAX(transaction_date) AS Maximum_date
from `workspace`.`default`.`case1`;


---3.Checking the different store location
SELECT DISTINCT store_location
from `workspace`.`default`.`case1`;

---4.Checking products sold at our stores- 9 different product category
SELECT DISTINCT product_category
from `workspace`.`default`.`case1`;


---5.Checking product type sold at our store-29 different product type
SELECT DISTINCT product_type
from `workspace`.`default`.`case1`;

---6.Checking product detail sold at our store- 80 different product details
SELECT DISTINCT product_detail
from `workspace`.`default`.`case1`;


---7.Checking for nulls in various columns
SELECT *
from `workspace`.`default`.`case1`
WHERE unit_price is null OR transaction_qty is null OR transaction_date is null;


---8.Checking lowest and highest unit price
SELECT 
      MIN(unit_price) AS lowest_unit_price,
      MAX(unit_price) AS Highest_unit_price
from `workspace`.`default`.`case1`;


---9.extracting the day and month names
SELECT transaction_date,
       Dayname(transaction_date) AS Day_name,
       Monthname(transaction_date) AS Month_name
from `workspace`.`default`.`case1`;


---10.
SELECT unit_price,
       transaction_qty,
       (unit_price*transaction_qty) AS revenue
from `workspace`.`default`.`case1`;  



---11.Combining Functions to get a clean enhanced data set
SELECT transaction_date,
       transaction_id,
       transaction_time,
       transaction_qty,
       store_id,
       store_location,
       product_id,
       product_category,
       product_type,
       product_detail,
-----Adding columns to enhance the table for better insights
---new column added 1
     Dayname(transaction_date) AS Day_name,
---new column added 2
     Monthname(transaction_date) AS Month_name,
---new column added 3
     Dayofmonth(transaction_date) AS Month_date,
---new column added 4-determining weekday/weekend
CASE
    WHEN Dayname(transaction_date) IN ('sunday','saturday') THEN 'weekend'
    ELSE 'weekday'
  END AS Day_classification,
---new column added 5-Time buckets
CASE
    
    WHEN date_format(transaction_time,'HH,MM,SS') BETWEEN '05:00:00' AND '09:59:59' THEN' 01.Rush hour'
   WHEN date_format(transaction_time,'HH,MM,SS') BETWEEN '10:00:00' AND '12:59:59' THEN' 02.Mid Morning'
   WHEN date_format(transaction_time,'HH,MM,SS') BETWEEN '13:00:00' AND '15:59:59' THEN' 03.Afternoon'
   WHEN date_format(transaction_time,'HH,MM,SS') BETWEEN '16:00:00' AND '18:59:59' THEN' 04.Rush Hour'
    ELSE '05.Night'
  END AS Time_classification,
---new column added 6-send bucket
CASE
     WHEN(transaction_qty*unit_price) <=50 THEN '01.Low Spend'
     WHEN(transaction_qty*unit_price) BETWEEN 51 AND 200 THEN '02.Medium Spender'
      WHEN(transaction_qty*unit_price) BETWEEN 201 AND 300 THEN '03.Spender'
    ELSE '04.Blesser'
    END AS Spend_bucket,
 ---new column added 7-revenue
     transaction_qty*unit_price AS Revenue
from `workspace`.`default`.`case1`;






  


















```
